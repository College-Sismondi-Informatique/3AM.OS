## Comment parcourir une liste
Il y a deux manières de parcourir une liste

### 1.Parcourir par position
- Utile lorsque l'on veut exploiter des informations de position
    
#### Exemple 1: Je veux copier une liste sauf le premier élément


```python
# Une liste contenant des valeurs
L_eleves = ["ADELE", "CELESTIN", "CEVDET", "ISKANDER", "LEO", 
    "LOUIS", "MARIUS", "MATHIS", "MIA", "NAYA", "PHUC-BAO", 
    "RENZO", "TAOHIDUL", "VICTOR", "VIKENTIOS"]

n = len(L_eleves) # la nombre d'éléments de la liste

# on parcourt la liste du second (élément à la position 1) au 
# dernier élément (élément à la position len(L_eleves)-1)
L_new = []
for k in range(1,n):
    L_new = L_new + [L_eleves[k]]
    # L_eleves[k] pour accéder à l'élément à la position k
    # Les crochets autour que ça devienne un tableau à 1 élément
    
print(L_new)
```

```python
# une alternative

# Une liste contenant des valeurs
L_eleves = ["ADELE", "CELESTIN", "CEVDET", "ISKANDER", "LEO", 
    "LOUIS", "MARIUS", "MATHIS", "MIA", "NAYA", "PHUC-BAO", 
    "RENZO", "TAOHIDUL", "VICTOR", "VIKENTIOS"]

n = len(L_eleves) # la nombre d'éléments de la liste


L_new = []
for k in range(0, n):
    if k>= 1:
        L_new = L_new + [L_eleves[k]]
    
print(L_new)
```

#### Exemple 2: Je veux inverser les éléments d'une liste

```python
# Une liste contenant des valeurs
L_eleves = ["ADELE", "CELESTIN", "CEVDET", "ISKANDER", "LEO", 
    "LOUIS", "MARIUS", "MATHIS", "MIA", "NAYA", "PHUC-BAO", 
    "RENZO", "TAOHIDUL", "VICTOR", "VIKENTIOS"]

n = len(L_eleves) # la nombre d'éléments de la liste

L_new = []
for k in range(0,n):
    L_new = L_new + [L_eleves[n-k-1]]
    # ou L_new = L_new + [L_eleves[-k-1]]
    # ou L_new = [L_eleves[k]] + L_new 
    
print(L_new)
```

#### Exemple 3:  Je veux créer une liste qui contient seulement la première moitié des éléments

```python
# Une liste contenant des valeurs
L_eleves = ["ADELE", "CELESTIN", "CEVDET", "ISKANDER", "LEO", 
    "LOUIS", "MARIUS", "MATHIS", "MIA", "NAYA", "PHUC-BAO", 
    "RENZO", "TAOHIDUL", "VICTOR", "VIKENTIOS"]

n = len(L_eleves) # la nombre d'éléments de la liste


L_new = []
for k in range(0,n):
    if k < n / 2:
        L_new = L_new + [L_eleves[k]]

    
print(L_new)
```

#### Exemple 4:  Je veux modifier les éléments d'une liste, par exemple ajouter un Titre

```python
# Une liste contenant des valeurs
L_eleves = ["ADELE", "CELESTIN", "CEVDET", "ISKANDER", "LEO", 
    "LOUIS", "MARIUS", "MATHIS", "MIA", "NAYA", "PHUC-BAO", 
    "RENZO", "TAOHIDUL", "VICTOR", "VIKENTIOS"]

n = len(L_eleves) # la nombre d'éléments de la liste

for k in range(0,n):
    L_eleves[k] = "Von " + L_eleves[k]

    
print(L_eleves)
```

### 2.Parcourir par élément
- Utile si on souhaite s'intéresser seulement aux éléments de la liste eux-mêmes
    
#### Exemple 1: Je veux copier une liste de notes qui contienne seulement les notes au dessus de la moyenne

```python
L_notes = [4.0, 5.5, 3.0, 6.0, 3.5, 2.0, 5.0, 5.5, 1.5, 4.5, 4.0]

L_new = []

for note in L_notes:
    if note >= 4.0:
        L_new = L_new + [note]# ici note est un élément de la liste
        
print(L_new)
```

**Remarque**
On peut toujours acomplir le même objectif avec un parcourt par position, mais la notation est plus chargée

```python
L_notes = [4.0, 5.5, 3.0, 6.0, 3.5, 2.0, 5.0, 5.5, 1.5, 4.5, 4.0]

L_new = []

for k in range(len(L_notes)):
    if L_notes[k] >= 4.0:
        L_new = L_new + [L_notes[k]]
        
print(L_new)
```
