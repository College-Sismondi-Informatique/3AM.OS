````python
from math import sqrt

def res_eq_2_degre(a, b, c):
    delta = b**2 - 4 * a * c
    if delta > 0 :
        x1 = (-b+sqrt(delta))/(2*a)
        x2 = (-b-sqrt(delta))/(2*a)
        S  = [x1, x2]
    elif delta == 0 :
        x = (-b)/(2*a)
        S = [x]
    else:
        S = []
        
    return S

sol = res_eq_2_degre(2, -8, 8)
print(sol)
sol = res_eq_2_degre(5, 10, -75)
print(sol)
sol = res_eq_2_degre(5, 10, 75)
print(sol)
````

1. **Intérêts simples :** On dispose d'une somme $S_0$. Chaque année ce placement rapporte des intérêts en fonction de la somme initiale.
   Par exemple avec $S_0 = 1000$ et des intérêts simples de $p = 10\%$, les intérêts sont de 100. Au bout d'un an $S_1 = 1100$, au bout de deux ans $S_2 = 1200$...
   
   **A faire:** programmer une fonction `interets_simples(S0, p, n)` qui renvoie la liste des sommes des $n$ premières années.
   Utiliser la fonction pour vérifier que  l'évolution d'une somme de 1000.- sur 3 ans à 10\% est 1100.0, 1200.0, 1300.0.

````python
# definition de la fonction
def interets_simples(S0, p, n):
    sommes = [S0]
    interet = S0 * (p / 100)
    somme_actuelle = S0
    for _ in range(n):
        somme_actuelle = somme_actuelle + interet
        sommes.append(somme_actuelle)
    return sommes

# Test de la fonction sur 3 ans avec 1000.- et 10% d'intérêts
l = interets_simples(1000, 10, 3)
print(l)
````

2. **Intérêts composés :** Une somme $S_0$ rapporte selon des intérêts composés. Cette fois les intérêts sont calculés chaque année sur la base de la somme de l'année précédente, selon la formule :
   $$I_{n+1} = S_n \times \frac{p}{100}$$
   
   *Programme une fonction *
   
   **A faire:** programmer une fonction `interets_composes(S0, p, n)` qui renvoie la liste des sommes des $n$ premières années.
   Utiliser la fonction pour vérifier que  l'évolution d'une somme de 1000.- sur 3 ans à 10\% est 1100.0, 1200.0, 1300.0.

````python
def interets_composes(S0, p, n):
    sommes = [S0]
    somme_actuelle = S0
    for _ in range(n):
        somme_actuelle += somme_actuelle * (p / 100)
        sommes.append(somme_actuelle)
    return sommes

# Test : Test de la fonction sur 3 ans avec 1000.- et 10% d'intérêts
l = interets_composes(1000, 10, 3)
print(l)
````

3. **Comparaison :** J'ai le choix entre un placement à intérêts simples de $10\%$ et un placement à intérêts composés de $7\%$. A partir que quelle moment le placement avec des intérêts composés de $7\%$ est-il plus avantageux?

   **A faire:** programmer une fonction `compare_interets(S0, p1, p2)` qui renvoie l'année à partir de laquelle il est plus avantajeux d'avoir des intérêts composés avec avec un taux `p2` que des intérêts simples avec un taux `p1`. 
   Utiliser la fonction pour répondre à la question de l'exercice.
   
   **Remarque**: vous pouvez utiliser une boucle `while`.

````python
def compare_interets(S0, p1, p2):
    S1 = S0  
    S2 = S0
    n=0
    while S1 >= S2:
        S1 = S1 + S0 * (p1 / 100)
        S2 = S2 + S2 * (p2 / 100)
        n = n + 1
        
    return n
    
# Test de la fonction 
annee = compare_interets(1000, 10, 7)
print(annee)
````

````python
def rotation(liste):
    n = len(liste)
    if n == 0:
        return []
    L = [liste[n-1]]
    for k in range(n-1):
         L = L + [liste[k]] # ou L.append(liste[k])
    
    return L

# Test des la fonction
l_rot = rotation([1, 2, 3, 4])
print(l_rot)
print(rotation(rotation([1, 2, 3, 4])))
````

````python
def supprimer_rang(liste, rang):
    res = []
    for k in range(len(liste)):
        if k != rang :
            res = res + [liste[k]]
    return res
# Test
l_supp = supprimer_rang([8, 7, 6, 5, 4], 2)
print(l_supp)
````

4. Programme une fonction `supprimer_element(liste, element)` renvoyant une liste qui contient tous les éléments sauf ceux égaux à l'élément spécifié.

```python
l_supp = supprimer_element([8, 7, 4, 6, 5, 4], 4)
print(l_supp)
```
> `[8, 7, 6, 5]`

````python
def supprimer_element(liste, element):
    res = []
    for e in liste:
        if e != element :
            res = res + [e]
    return res

# Test
l_supp = supprimer_element([8, 7, 4, 6, 5, 4], 4)
print(l_supp)
````

````python
def minimum(l):
    i_min = 0
    for k in range(len(l)):
        if l[k] < l[i_min]:
            i_min = k
    return i_min


i_min = minimum([13, 11, 7, 16, 6, 8, 12, 6])
print(i_min)
````

##### Algorithme :
- **Entrée :** une liste $\ell$ de $n$ nombres

Pour vérifier qu'un nombre `d` est un diviseur de `n` il suffit de vérifier si le reste de la division entière de `d` par `n` vaut 0 (en python : `n % d == 0`).

````python
def diviseurs(n):
    D = []
    for d in range(1,n+1):
        if n % d == 0:
            D = D + [d]
    return D

# Test pour n = 420
Div = diviseurs(420)
print(Div)
````

3*. **Facteurs premiers :** Programme une fonction `facteurs_premiers(n)` qui renvoie la liste de tous les facteurs premiers d'un entier $n \ge 2$.

```python
l_facteurs = facteurs_premiers(12936)
print(l_facteurs)
```
> `[2, 2, 2, 3, 7, 7, 11]`


````python
def facteurs_premiers(n):
    facteurs = []
    d = 2 # premiers diviseurs premiers possibles
    while d <= n:
        if n % d == 0:
            facteurs.append(d)
            n = n // d
        else:
            d += 1
    return facteurs


# Test pour n = 12936
l_facteurs = facteurs_premiers(12936)
print(l_facteurs)
````

Pour obtenir un tableau de valeurs plus précis, on prèfère untiliser une boucles:

````python
import matplotlib.pyplot as plt

x_min, x_max = 2, 9 # Les bornes de l'intervalle sur lequel
                    # tracer le graphique
deltaX = 0.1 # la distance entre deux valeurs du tableau de valeurs

def f(x):
    return x**2

liste_x = []
liste_y = []
x = x_min
while x <= x_max:
    liste_x = liste_x + [x]
    liste_y = liste_y + [f(x)]
    x = x + deltaX
    

plt.figure()
plt.plot(liste_x, liste_y, label="$y=x^2$", color="red")
plt.grid(True)
plt.legend()
plt.show()

````

## Activité 5 : Tir balistique

> **Objectifs :** Visualiser le tir d'un boulet de canon.

L'équation de la trajectoire est donnée par :
$$y(x) = -\frac{1}{2} g \frac{1}{v^2 \cos^2(\alpha)} x^2 + \tan(\alpha) x$$
où :
- $\alpha$ est l'angle du tir
- $v$ est la vitesse initiale
- $g = 9.81$ est la constante de gravitation

1. **Fonction `tir_parabolique(x, v, alpha)`** :

````python
import math

def tir_parabolique(x, v, alpha_degres):
    g = 9.81
    alpha_rad = (2 * math.pi / 360) * alpha_degres
    y = -0.5 * g * (1 / (v**2 * math.cos(alpha_rad)**2)) * (x**2) + math.tan(alpha_rad) * x
    return y
````

2. **Fonction `liste_trajectoire(xmax, n, v, alpha)`** :

````python
def liste_trajectoire(xmax, n, v, alpha):
    liste_y = []
    liste_x = []
    for i in range(n + 1):
        xi = i * (xmax / n)
        yi = tir_parabolique(xi, v, alpha)
        liste_x.append(xi)
        liste_y.append(yi)
    return liste_x, liste_y
````

3. **Affichage des trajectoires pour différentes valeurs de l'angle $\alpha$ :**

````python
v = 50
xmax = 270
n = 100

plt.figure(figsize=(9, 5))
for alpha in [30, 45, 60, 75]:
    lx, ly = liste_trajectoire(xmax, n, v, alpha)
    plt.plot(lx, ly, label=f"α = {alpha}°")

plt.axhline(0, color='black', linewidth=0.8, linestyle='--')
plt.ylim(-100, 100)
plt.grid(True)
plt.xlabel("x")
plt.ylabel("y(x)")
plt.title("Trajectoires balistiques")
plt.legend()
plt.show()
````
