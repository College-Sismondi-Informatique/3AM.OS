````python
from math import sqrt

def res_eq_2_degre(a, b, c):
    delta = b**2 - 4 * a * c
    if delta > 0 :
        x1 = (-b+sqrt(delta))/(2*a)
        x2 = (-b-sqrt(delta))/(2*a)
        S  = [x1, x2]
    elif delta == 0 :
        x = (-b)/(2*a)
        S = [x]
    else:
        S = []
        
    return S

sol = res_eq_2_degre(2, -8, 8)
print(sol)
sol = res_eq_2_degre(5, 10, -75)
print(sol)
sol = res_eq_2_degre(5, 10, 75)
print(sol)
````

1. **Intérêts simples :** On dispose d'une somme $S_0$. Chaque année ce placement rapporte des intérêts en fonction de la somme initiale.
   Par exemple avec $S_0 = 1000$ et des intérêts simples de $p = 10\%$, les intérêts sont de 100. Au bout d'un an $S_1 = 1100$, au bout de deux ans $S_2 = 1200$...
   
   **A faire:** programmer une fonction `interets_simples(S0, p, n)` qui renvoie la liste des sommes des $n$ premières années.
   Utiliser la fonction pour vérifier que  l'évolution d'une somme de 1000.- sur 3 ans à 10\% est 1100.0, 1200.0, 1300.0.

````python
# definition de la fonction
def interets_simples(S0, p, n):
    sommes = [S0]
    interet = S0 * (p / 100)
    somme_actuelle = S0
    for _ in range(n):
        somme_actuelle = somme_actuelle + interet
        sommes.append(somme_actuelle)
    return sommes

# Test de la fonction sur 3 ans avec 1000.- et 10% d'intérêts
l = interets_simples(1000, 10, 3)
print(l)
````

2. **Intérêts composés :** Une somme $S_0$ rapporte selon des intérêts composés. Cette fois les intérêts sont calculés chaque année sur la base de la somme de l'année précédente, selon la formule :
   $$I_{n+1} = S_n \times \frac{p}{100}$$
   
   *Programme une fonction *
   
   **A faire:** programmer une fonction `interets_composes(S0, p, n)` qui renvoie la liste des sommes des $n$ premières années.
   Utiliser la fonction pour vérifier que  l'évolution d'une somme de 1000.- sur 3 ans à 10\% est 1100.0, 1210.0, 1331.0.

````python
def interets_composes(S0, p, n):
    sommes = [S0]
    somme_actuelle = S0
    for _ in range(n):
        somme_actuelle += somme_actuelle * (p / 100)
        sommes.append(somme_actuelle)
    return sommes

# Test : Test de la fonction sur 3 ans avec 1000.- et 10% d'intérêts
l = interets_composes(1000, 10, 3)
print(l)
````

3. **Comparaison :** J'ai le choix entre un placement à intérêts simples de $10\%$ et un placement à intérêts composés de $7\%$. A partir que quelle moment le placement avec des intérêts composés de $7\%$ est-il plus avantageux?

   **A faire:** programmer une fonction `compare_interets(S0, p1, p2)` qui renvoie l'année à partir de laquelle il est plus avantajeux d'avoir des intérêts composés avec avec un taux `p2` que des intérêts simples avec un taux `p1`. 
   Utiliser la fonction pour répondre à la question de l'exercice.
   
   **Remarque**: vous pouvez utiliser une boucle `while`.

````python
def compare_interets(S0, p1, p2):
    S1 = S0  
    S2 = S0
    n=0
    while S1 >= S2:
        S1 = S1 + S0 * (p1 / 100)
        S2 = S2 + S2 * (p2 / 100)
        n = n + 1
        
    return n
    
# Test de la fonction 
annee = compare_interets(1000, 10, 7)
print(annee)
````

````python
def supprimer_rang(liste, rang):
    res = []
    for k in range(len(liste)):
        if k != rang :
            res = res + [liste[k]]
    return res
# Test
l_supp = supprimer_rang([8, 7, 6, 5, 4], 2)
print(l_supp)
````

4. Programme une fonction `supprimer_element(liste, element)` renvoyant une liste qui contient tous les éléments sauf ceux égaux à l'élément spécifié.

```python
l_supp = supprimer_element([8, 7, 4, 6, 5, 4], 4)
print(l_supp)
```
> `[8, 7, 6, 5]`

````python
def supprimer_element(liste, element):
    res = []
    for e in liste:
        if e != element :
            res = res + [e]
    return res

# Test
l_supp = supprimer_element([8, 7, 4, 6, 5, 4], 4)
print(l_supp)
````

````python
def minimum(l):
    i_min = 0
    for k in range(len(l)):
        if l[k] < l[i_min]:
            i_min = k
    return i_min


i_min = minimum([13, 11, 7, 16, 6, 8, 12, 6])
print(i_min)
````

##### Algorithme :
- **Entrée :** une liste $\ell$ de $n$ nombres

  - supprimer l'élément à la position $i$ de  $\ell$

````python
def facteurs_premiers(n):
    facteurs = []
    d = 2 # premiers diviseurs premiers possibles
    while d <= n:
        if n % d == 0:
            facteurs.append(d)
            n = n // d
        else:
            d += 1
    return facteurs


# Test pour n = 12936
l_facteurs = facteurs_premiers(12936)
print(l_facteurs)
````

````python
from math import tan, pi, cos

def tir_parabolique(x,v,alpha):
    alpha = 2*pi*alpha/360
    return -0.5*9.81*x**2/ (v**2 * cos(alpha)**2) + tan(alpha)*x
    
print(tir_parabolique(20, 500, 30))
````

Un nombre de Kaprekar est un nombre dont le carré peut-être coupé en une partie gauche et une partie droite (non-nulle) telles que la somme des deux parties donne le nombre initial.

1. Ecrire une fonction `est_kaprekar(n)` qui renvoie `True` si l'entier `n` est un nombre de Kaprekarle et `False` sinon.

   Vérifier que 2223 et 297 sont des nombre de Kaprekar mais pas 785.

<summary style="border-left:3px solid #3c763d; border-radius:2pt; width:100%; color:#3c763d; padding:6px; background-color: #dff0d8">

</summary>  

<div style="border-left:3px solid #3c763d; border-radius:2pt; color:#3c763d; padding:6px; background-color: #eff0e8"> 

Pour connaitre le nombre de chiffres il suffit de calculer le $\log_{10}$ et de faire un arrondi supérieur avec la fonction `ceil(n)` du module `math`.
    
```python
from math import log, ceil
r = log(47862, 10)
L = ceil(r)   
print(L)
```
> `5`

Pour connaitre les $k$ premiers chiffres, on peut faire une division entière avec $10^k$.
    
</div>
</details>

````python
import math
def est_kaprekar(n):
    m = n**2
    L = math.ceil(log(m+0.1, 10))# nombre de chiffres
    for k in range(1,L):
        a = m // 10**k
        b = m - a*10**k
        if n == a+b and b!= 0:
            return True
        
    return False

print(297, "est-il un nombre de kaprekar?",est_kaprekar(297))
print(2223, "est-il un nombre de kaprekar?",est_kaprekar(2223))
print(785, "est-il un nombre de kaprekar?",est_kaprekar(785))
````

````python
def puissance(liste, k):
    return [n**k for n in liste]

print(puissance([1,2,3,4,5],3))
````

3. Programme une fonction `non_zero(liste)` qui renvoie la liste de tous les éléments non nuls. Par exemple `non_zero([1,0,2,3,0,4,5,0])` renvoie `[1,2,3,4,5]`.


````python
def non_zero(liste):
    return [n for n in liste if n != 0]

print(non_zero([1,0,2,3,0,4,5,0]))
````

4. Programme une fonction `pairs(liste)` qui renvoie la liste de tous les éléments pairs. Par exemple `pairs([1,0,2,3,0,4,5,0])` renvoie `[0,2,0,4,0]`.

````python
def pairs(liste):
    return [n for n in liste if n %2 == 0]

print(pairs([1,0,2,3,0,4,5,0]))
````
