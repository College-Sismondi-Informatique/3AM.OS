<h3 style="color:teal;background-color:azure;" > <i class="fa fa-pencil" aria-hidden="true"> </i> &nbsp; Exercice 1.44 (Algorithme d'Euclide étendu) </h3>


### Retour sur l'algorithme d'Euclide
L'algorithme d'Euclide, pour trouver le pgcd de deux nombres entiers $a$ et $b$, consiste à effectuer des divisions euclidiennes successives. En posant $r_0=a$ et $r_1 = b$, on obtient :

$$\begin{aligned}
r_0 &= r_1 \cdot q_1 + r_2 && (r_2 \neq 0) \\
r_1 &= r_2 \cdot q_2 + r_3 && (r_3 \neq 0) \\
&\;\;\vdots \\
r_{n-2} &= r_{n-1} \cdot q_{n-1} + r_n && (r_n \neq 0) \\
r_{n-1} &= r_n \cdot q_n && (r_{n+1} = 0)
\end{aligned}$$

La proposition 1.3 du cours nous assure que

$$\text{pgcd}(a; b) = \text{pgcd}(r_0; r_1) = \text{pgcd}(r_i; r_{i+1})$$

et donc que le pgcd est $r_n$, le dernier reste non nul, obtenu juste avant que $r_{n+1} = 0$.

On peut récrire les égalités précédentes en isolant les restes :

$$\begin{aligned}
r_0 &= a  && \ \\
r_1 &= b  && \ \\
r_0 &= r_1 \cdot q_1 + r_2 &\Leftrightarrow &\quad r_2 = r_0-r_1 \cdot q_1\\
r_1 &= r_2 \cdot q_2 + r_3 &\Leftrightarrow &\quad r_3 = r_1-r_2 \cdot q_2\\
&\;\;\vdots \\
r_{n-2} &= r_{n-1} \cdot q_{n-1} + r_n &\Leftrightarrow &\quad r_n = r_{n-2}-r_{n-1} \cdot q_{n-1}\\
r_{n-1} &= r_n \cdot q_n &\Leftrightarrow &\quad  r_{n+1} = 0
\end{aligned}$$

On obtient donc la formule de récurrence suivante sur les restes :

$$
r_{i+1} = r_{i-1} - q_i r_i
$$

où $q_i$ représente le quotient de la division de $r_{i-1}$ par $r_{i}$.

### L'algorithme d'Euclide étendu 
Le but de **l'algorithme d'Euclide étendu** est de calculer les coefficients de Bézout $u$ et $v$ tels que $ua + vb = \text{pgcd}(a,b)$. Dans les exemples du cours, nous procédons en « remontant » l'algorithme d'Euclide afin de trouver les coefficients de Bézout.

L'idée de **l'algorithme d'Euclide étendu** est de montrer qu'il est possible de trouver une formule récursive qui permet de calculer les coefficients exprimant chaque reste $r_i$ comme une combinaison linéaire de $a$ et $b$ : $r_i=u_i\cdot a + v_i \cdot b$.

Pour les deux premiers restes, c'est évident :

$$\begin{aligned}
r_0 &= 1\cdot a + 0\cdot b = u_0\cdot a + v_0 \cdot b && \ \\
r_1 &= 0\cdot a + 1\cdot b = u_1\cdot a + v_1 \cdot b && \ \\
\end{aligned}$$

Ce qui donne $u_0 = 1$, $v_0 = 0$, $u_1 = 0$ et $v_1 = 1$. Pour la suite, il suffit d'exploiter la formule $r_{i+1} = r_{i-1} - q_i r_i$ :

$$
\begin{aligned}
r_{i+1} &= r_{i-1} - q_i r_i \\
        &= \left(u_{i-1}\cdot a + v_{i-1} \cdot b \right) - q_i \left(u_{i}\cdot a + v_{i} \cdot b \right) \\ 
        &= \underbrace{\left(u_{i-1} - q_i u_{i} \right)}_{u_{i+1}}\cdot a
           + \underbrace{\left(v_{i-1}-q_i v_{i}\right)}_{v_{i+1}}\cdot b \\
        &= u_{i+1}\cdot a + v_{i+1}\cdot b
\end{aligned}
$$

avec $u_{i+1} = u_{i-1} - q_i u_{i}$ et $v_{i+1} = v_{i-1}-q_i v_{i}$. L'algorithme se termine lorsque $r_{n+1}=0$ et les coefficients $u_{n}$ et $v_{n}$ sont les coefficients de Bézout.

### Conclusion

On effectue l'algorithme d'Euclide et les coefficients de Bézout s'obtiennent en calculant successivement :

$$\begin{cases} u_{i+1} = u_{i-1} - q_i \cdot u_i \\ v_{i+1} = v_{i-1} - q_i \cdot v_i \end{cases} \quad \text{avec} \quad \begin{cases} u_0 = 1 \\ v_0 = 0 \end{cases} \quad \text{et} \quad \begin{cases} u_1 = 0 \\ v_1 = 1 \end{cases}$$

où $q_i$ représente le quotient obtenu à chaque étape de l'algorithme d'Euclide.

## Exercice : le crible d'Ératosthène

Le [crible d'Ératosthène](https://fr.wikipedia.org/wiki/Crible_d%27%C3%89ratosth%C3%A8ne) est un algorithme qui permet de trouver tous les nombres premiers inférieurs ou égaux à un entier $N$. Le principe : on écrit tous les entiers de $2$ à $N$, puis on prend le plus petit entier non barré $p$ (il est premier), on barre tous ses multiples, et on recommence avec le prochain entier non barré.

1. Écrire une fonction `crible(N)` qui renvoie la liste des nombres premiers inférieurs ou égaux à $N$ en suivant la méthode ci-dessous :
    - créer une liste de booléens `est_p` de taille $N+1$, initialisée à `True` ;
    - mettre `est_p[0]` et `est_p[1]` à `False` (car 0 et 1 ne sont pas premiers);
    - pour chaque $p$ de $2$ à $N$, si `est_p[p]` vaut `True`, mettre à `False` toutes les cases `est_p[m]` où $m$ est un multiple de $p$ ;
