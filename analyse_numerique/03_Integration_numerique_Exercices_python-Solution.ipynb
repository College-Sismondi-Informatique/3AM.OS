### Exercice 37

a) Ecrire une fonction `rectangle(g,a,b,n)` qui retourne l’approximation de $\int_{a}^{b}{g\left( c \right)dx}$ d’ordre n.


```python
def rectangle(f,a,b,n):
    h = (b-a)/n
    somme = 0
    for k in range(n):
        somme = somme + f(a+k*h)
    somme = h * somme
    return somme
```

- $g(x)=\sin \left( x \right);a=0;b=\pi ;n=1000$ 

```python
def f(x):
    return x**2

a, b = 0, 1

for n in [3,5, 1000]:
    print(f"L'intégrale définie entre {a} et {b} de f:x↦x² estimée avec la méthode des rectangles pour {n} itération vaut",
      rectangle(f,a,b,n))
    print()
```

```python
from math import pi, sin

def g(x):
    return sin(x)

a, b = 0, pi
n = 1000
print(f"L'intégrale définie entre {a} et {b} de g:x↦sin(x) estimée avec la méthode des rectangle pour {n} itération vaut""",
      rectangle(g,a,b,n))
```

### Exercice 39

a) Ecrire une fonction `trapeze(g,a,b,n)` qui retourne l’approximation de $\int_{a}^{b}{g\left( c \right)dx}$ d’ordre n en utilisant la méthode des trapèzes.


```python
def trapeze(f,a,b,n):
    h = (b-a)/n
    somme = f(a)/2 + f(b)/2
    for k in range(1,n):
        somme = somme + f(a+k*h)
    somme = h * somme
    return somme
```

### Exercice 41

Soit $f$ une fonction définie par $f(x)=\frac{\sin (x)}{x}$ qui ne possède pas de primitive s’exprimant 
à l’aide de fonctions simples. 
On peut toutefois obtenir une primitive de $f$ par :  $F(x)=\int\limits_{\frac{\pi }{4}}^{x}{f(t)dt}$ pour tout $x\in \left[ \frac{\pi }{4};4\pi  \right]$.

a) Utiliser la fonction `trapeze(f,a,b,n)` avec $n=100$ pour approximer les images de :  $\frac{\pi }{4};\frac{\pi }{2};\frac{3\pi }{4};\pi ;\frac{5\pi }{4};\frac{3\pi }{2};\frac{7\pi }{4};2\pi ;...;4\pi $ par la fonction primitive $F$  et établir un tableau des valeurs de la fonction $f$ et $F$.

*Indication : Stocker les pré-images et les images dans des listes `X`, `Y_f` et `Y_F`. Pour obtenir une liste de `n` valeurs entres deux bornes `a` et `b` on peut utiliser la fonction `linspace(a, b, n)` du module `numpy`.*


```python
from math import pi, sin
from numpy import linspace

n = 100

def f(x):
    return sin(x)/x

def F(x):
    return trapeze(f,pi/4, x, n)

X = linspace(pi/4, 4*pi, 50)
Y_f = [f(x) for x in X]
Y_F = [F(x) for x in X]

print(X[:4])
print(Y_f[:4])
print(Y_F[:4])
```

b) Tracer à l’aide du module `matplotlib` le graphique de la fonction $f$  et $F$ dans un même repère sur l’intervalle $\left[ \frac{\pi }{4};4\pi  \right]$.  

```python
import matplotlib.pyplot as plt

plt.plot(X, Y_f, 'k-',linewidth=3,label='noir : f')
plt.plot(X, Y_F, 'r-',linewidth=3,label='rouge : F, la primitive de f')

plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()

plt.show()
```

### Exercice 42

La fonction accélération est définie comme le taux de variation instantané de la vitesse, 
c’est-à-dire comme la dérivée de la fonction vitesse, ce que nous pouvons encore formuler : 
«  la fonction vitesse est une primitive de la fonction accélération » .

- $a(t)$ est l’accélération instantanée à l’instant $t$ et on a $a(t)={{\left( v(t) \right)}^{'}}$

 - $v(t)$ est la vitesse instantanée à l’instant $t$ et on  a $v({{t}_{1}})-v({{t}_{0}})=\int\limits_{{{t}_{0}}}^{{{t}_{1}}}{a(t)\ dt}$

On rappelle que la formule initiale de la méthode du trapèze  

${{T}_{n}}=\sum\limits_{k=0}^{n-1}{\frac{1}{2}\cdot \left[ f\left( a+k\cdot \frac{b-a}{n} \right)+f\left( a+\left( k+1 \right)\cdot \frac{b-a}{n} \right) \right]}\cdot \frac{b-a}{n}$

calcule l’air sous la courbe de f où le terme de la somme représente l’aire sous la courbe entre deux valeurs successive  de la subdivision qui est approximé comme l’aire d’un trapèze. Ici si on prend $a(t)$ à la place de $f(x)$ on peut voir ce terme comme une approximation de $\int_{{{t}_{0}}}^{{{t}_{1}}}{a(t)\ dt}$.


#### Enoncé

L’ordinateur de bord d’une « formule un » à enregistré les mesures suivantes :

| Temps [s] | 9  | 10 | 11 | 12 | 13 | 14 | 15 | 16 | 17 |
| --------- | -- | -- | -- | -- | -- | -- | -- | -- | -- |
| Accélération [m/s2] | 12 | 10.4 | 9.8 | 8 | -5 | -6.2 | 9.5 | 3.3 | 7.5 |

Sachant que la vitesse instantanée de la voiture au temps t = 9 [s] était de 30 [m/s] , écrire un programme python qui estime la vitesse instantanée de la voiture pour chaque mesure.

Indications : - Définir une liste `T = [9,10,11,12,13,14,15,16,17]`et une liste `A=[12,10.4,9.8,8,-5,-6.2,9.5,3.3,7.5]`
- Utiliser les instructions : len(A), A[0], A[1], etc.


```python
T = [9,10,11,12,13,14,15,16,17]
A=[12,10.4,9.8,8,-5,-6.2,9.5,3.3,7.5]
V= [30]

h=1 # le temps entre deux mesures

for i in range(0,len(A)-1):
    aire_trapeze=(A[i]+A[i+1])*(h/2)
    new_v = V[i] + aire_trapeze
    V.append(new_v)

print("Vitesses en [m/s] :", V , "\n")

V_km = [v*3600/1000 for v in V]
print("Vitesses en [km/h] :", V_km)
```

### Exercice 43

a) Ecrire une fonction simpson(f,a,b,n) qui retourne l’approximation de $\int_{a}^{b}{f\left( c \right)dx}$ d’ordre n en utilisant la méthode des trapèzes.


```python
def simpson(f,a,b,n):
    h = (b-a)/n
    Si = 0
    Sp = 0
    if n == 2:
        return (h/3)*(f(a)+4*(f(a+h)+Si)+2*Sp+f(b))
    else:
        d=n//2
        for k in range(1,d):
            Si = Si + f(a+(2*k+1)*h)
            Sp = Sp + f(a+2*k*h)
        return (h/3) * (f(a) + 4 * (f(a+h) + Si) + 2 * Sp + f(b))
```

```python
def f(x):
    return x*x

simpson(f,0,1,10)
```

Pour chacune des trois méthodes vues en cours (rectangles, trapèzes et Simpson), déterminer le nombre minimum de subdivisions n de l'intervalle $\left[ 1;2 \right]$  pour être certain d’avoir une erreur inférieure ou égale à ${{10}^{-4}}$en résolvant une inéquation.

### Solution:

**Méthode des Rectangles :**

$\begin{align}
  & \left| {{R}_{n}}-\int\limits_{a}^{b}{f(t)dt} \right|\le \frac{{{(b-a)}^{2}}}{2n}\cdot \underset{t\in \left[ a;b \right]}{\mathop{\max }}\,\left| f'(t) \right|\le {{10}^{-4}} \\ 
 & \Leftrightarrow \left| {{R}_{n}}-\int\limits_{1}^{2}{\frac{ln(t)}{t}dt} \right|\le \frac{1}{2n}\cdot \underbrace{\underset{t\in \left[ 1;2 \right]}{\mathop{\max }}\,\left| \frac{1-\ln (t)}{{{t}^{2}}} \right|}_{=1}\le {{10}^{-4}} \\ 
 & \Leftrightarrow \frac{1}{2n}\le {{10}^{-4}}\Leftrightarrow \frac{1}{{{10}^{-4}}}\le 2n\Leftrightarrow {{10}^{4}}\le 2n\Leftrightarrow n\ge \frac{10'000}{2}\Leftrightarrow \left. {\underline {\, 
 \left. {\underline {\, 
 n\ge 5000 \,}}\! \right|  \,}}\! \right|  \\ 
\end{align}$

**Méthode des Trapèzes :**
  
$\begin{align}
  & \left| {{T}_{n}}-\int\limits_{a}^{b}{f(t)dt} \right|\le \frac{{{(b-a)}^{3}}}{12{{n}^{2}}}\cdot \underset{t\in \left[ a;b \right]}{\mathop{\max }}\,\left| f''(t) \right|\le {{10}^{-4}} \\ 
 & \Leftrightarrow \left| {{T}_{n}}-\int\limits_{1}^{2}{\frac{ln(t)}{t}dt} \right|\le \frac{1}{12{{n}^{2}}}\cdot \underbrace{\underset{t\in \left[ 1;2 \right]}{\mathop{\max }}\,\left| \frac{-3+2\ln (t)}{{{t}^{3}}} \right|}_{=3}\le {{10}^{-4}} \\ 
 & \Leftrightarrow \frac{1}{4{{n}^{2}}}\le {{10}^{-4}}\Leftrightarrow \frac{1}{{{10}^{-4}}}\le 4{{n}^{2}}\Leftrightarrow {{10}^{4}}\le 4{{n}^{2}}\Leftrightarrow 2500\le {{n}^{2}}\Leftrightarrow \left. {\underline {\, 
 \left. {\underline {\, 
 n\ge 50 \,}}\! \right|  \,}}\! \right|  \\ 
\end{align}$



**Méthode de Simpson :**

$\begin{align}
  & \left| {{S}_{n}}-\int\limits_{a}^{b}{f(t)dt} \right|\le \frac{{{(b-a)}^{5}}}{180{{n}^{4}}}\cdot \underset{t\in \left[ a;b \right]}{\mathop{\max }}\,\left| {{f}^{(4)}}(t) \right|\le {{10}^{-4}} \\ 
 & \Leftrightarrow \left| {{S}_{n}}-\int\limits_{1}^{2}{\frac{ln(t)}{t}dt} \right|\le \frac{{{1}^{5}}}{180{{n}^{4}}}\underbrace{\underset{t\in \left[ 1;2 \right]}{\mathop{\max }}\,\left| \frac{-50+24\ln (t)}{{{t}^{5}}} \right|}_{=50}\le {{10}^{-4}} \\

 \left. {\underline {\, 
 n\ge 8 \,}}\! \right|  \,}}\! \right|  \\ 
\end{align}$
