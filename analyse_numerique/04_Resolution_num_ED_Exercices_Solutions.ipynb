```python
# Fonction f
def f(x,y):
    return y

# Méthode d'Euler
def euler(f,x0,y0,b,n):
    h=(b-x0)/n
    X, Y = [x0], [y0]
    for _ in range(n):
        x1 = x0 + h
        y1 = y0 + h * f(x0,y0)
        X = X + [x1]
        Y = Y + [y1]
        x0 = x1
        y0 = y1
    return X,Y
```

### Exercice 48

**Loi de refroidissement de Newton**

Si un corps est placé dans un milieu ambiant dont la température $T_a$ est constante, la température $T$ (en Celsius) du corps au temps $t$ (en minute) est donnée par la loi de refroidissement de Newton :

>  La vitesse de refroidissement du corps est proportionnelle à la différence des températures du corps et du milieu ambiant.

Cette observation peut se modéliser avec l’équation différentielle suivante : 

$\frac{dT}{dt}=k\cdot \left( T-{{T}_{a}} \right)$  

Remarques : 
- $T$ est fonction du temps $t$  d’où \[T=T\left( t \right)\]
- $k$ est le taux de croissance/décroissance (constante). 

**Enoncé**

**a)** Montrer que  $T={{T}_{a}}+({{T}_{0}}-{{T}_{a}})\cdot {{e}^{kt}}$  est une solution particulière de l’E.D 
    avec comme C.I.  $T\left( 0 \right)={{T}_{0}}$ .
    Remarque : Cette E.D. peut-être résolue analytiquement par séparation des variables.



**Solution**
- $\frac{dT}{dt}={{\left( {{T}_{a}}+({{T}_{0}}-{{T}_{a}})\cdot {{e}^{kt}} \right)}^{'}}=0+({{T}_{0}}-{{T}_{a}})\cdot {{e}^{kt}}\cdot k=({{T}_{0}}-{{T}_{a}})\cdot {{e}^{kt}}\cdot k$
- $k\cdot \left( T-{{T}_{a}} \right)=k\cdot \left( {{T}_{a}}+({{T}_{0}}-{{T}_{a}})\cdot {{e}^{kt}}-{{T}_{a}} \right)=({{T}_{0}}-{{T}_{a}})\cdot {{e}^{kt}}\cdot k$
- $T\left( 0 \right)={{T}_{a}}+({{T}_{0}}-{{T}_{a}})\cdot {{e}^{0\cdot t}}={{T}_{a}}+{{T}_{0}}-{{T}_{a}}={{T}_{0}}$


**b)** Dans une pièce dont la température est de $T_a=20°$, se trouve une bouilloire remplie d'eau dont la température initiale est de $T_0=100°$.  Pour l’eau la constante $k=-0.0192$.
     
1) Ecrire un programme en Python 3 qui affiche en sortie sur l’intervalle $\left[ 0;300 \right]$ et dans le même repère :
 - la solution exacte de l’équation différentielle qui est $T=T_a+(T_0-T_a)\cdot e^{kt}$ (en trait plein noir).
 - la ligne brisée résultat de la méthode d’Euler avec $n={{2}^{4}}=16$ subdivisions (en trait-tillé rouge).

*Indication: S'inspirer de l'exercice précédant*

```python
# Importation modules
from math import exp

import matplotlib.pyplot as plt

# Paramètres (constantes)
x0,y0,b,n,Ta,k=0,100,300,2**4,20,-0.0192

# Solution exacte y(x)
def y(x):
    return Ta+(y0-Ta) * exp(k*x)

def tab_val(y,x0,b,n):
    X, Y = [x0], [y(x0)]
    h=(b-x0)/n
    while x0 <= b:
        X.append(x0)
        Y.append(y(x0))
        x0 = x0 + h
    return X,Y

# Méthode d’Euler
def f(x,y):
    return k*(y-Ta)

def euler(f,x0,y0,b,n):
    h=(b-x0)/n
    X, Y = [x0], [y0]
    for k in range(1,n+1):
        x1=x0+h
        y1=y0+h*f(x0,y0)
        X.append(x1)
        Y.append(y1)
        x0=x1
        y0=y1
    return X,Y

# Tableau de valeurs de la solution exacte
X, Y = tab_val(y,x0,b,n)

# Tableau de valeurs de la solution approchée
X_euler, Y_euler = euler(f,x0,y0,b,n)

# Graphique de y(x) et  yn
plt.figure()
plt.grid(True)
plt.xlabel("temps")
plt.ylabel("Température")
plt.xticks(range(0, b+1, 20), fontsize=7)
plt.yticks(range(0, y0+1, 10), fontsize=7)
p1,= plt.plot(X, Y,linestyle='-',color='black', marker='o', linewidth=4)
p2,= plt.plot(X_euler, Y_euler,linestyle='--',color='red', marker='o', linewidth=2)
plt.legend([p1,p2],['T(t) solution exacte','Tn(t) solution Euler'])
plt.show()
```

3) Au bout de combien de temps la température de l'eau ne sera-t-elle plus que de ${{40}^{\circ }}$ ?


**Solution**

$T=40{}^\circ \to t\cong 72\min \cong 1h12\min $ (lecture graphique avec la solution exacte).

#### Remarque

Résolution analytique de l’E.D. avec la méthode de séparation des variables.

$\frac{dT}{dt}=k\cdot \left( T-{{T}_{a}} \right)\Leftrightarrow \frac{dT}{T-{{T}_{a}}}=kdt\Leftrightarrow \int{\frac{1}{T-{{T}_{a}}}dT}=k\int{dt}\Leftrightarrow \ln \left| T-{{T}_{a}} \right|=kt+CC,k\in \mathbb{R}$

$\Leftrightarrow {{e}^{\ln \left| T-{{T}_{a}} \right|}}={{e}^{kt+C}}\Leftrightarrow \left| T-{{T}_{a}} \right|={{e}^{C}}\cdot {{e}^{kt}}\Leftrightarrow T-{{T}_{a}}=\underbrace{\pm {{e}^{C}}}_{B}\cdot {{e}^{kt}}\Leftrightarrow T(t)={{T}_{a}}+B\cdot {{e}^{kt}}B,k\in \mathbb{R}$

Condition initiale : $T(0)={{T}_{a}}+B{{e}^{k0}}={{T}_{0}}\Rightarrow {{T}_{a}}+B={{T}_{0}}\Rightarrow B={{T}_{0}}-{{T}_{a}}$

Donc la solution particulière est : $T(t)={{T}_{a}}+\left( {{T}_{0}}-{{T}_{a}} \right)\cdot {{e}^{kt}}$ avec $k\in \mathbb{R}$


```python
import numpy as np
import matplotlib.pyplot as plt
from math import exp

# 1. Paramètres du problème
Ta = 20      # Température ambiante (°C)
T0 = 100      # Température initiale du corps (°C)
k = -0.0192     # Constante de refroidissement (négative car la température baisse)

# 2. Solution exacte de l'ED : dT/dt = k * (T - Ta)
def T(t):
    return Ta + (T0 - Ta) * exp(k*t) 

# 3. Création du champ de vecteurs (Grille de points)
t_grid = np.linspace(0, 60, 20)  # Temps de 0 à 60 min
T_grid = np.linspace(0, 100, 20) # Température de 0 à 100 °C
tt, TT = np.meshgrid(t_grid, T_grid)
DT = k * (TT - Ta) # Calcul de la pente en chaque point
# Pour le champ de vecteurs, on fixe la composante temporelle à 1
Dt = np.ones_like(DT) 

# 4. Calcul de la solution exacte (courbe)
t_sol = np.linspace(0, 60, 100)
T_sol = [T(t) for t in t_sol]

# 5. Graphique
plt.figure(figsize=(10, 6))

# Affichage du champ de vecteurs (normalisé pour la lisibilité)
plt.quiver(tt, TT, Dt, DT, color='gray', alpha=0.4, label="Champ de direction (dT/dt)")

# Affichage de la solution
plt.plot(t_sol, T_sol, 'r', linewidth=2, label=f"Solution T(t) (Initial: {T0}°C)")

# Ligne de la température ambiante
plt.axhline(Ta, color='blue', linestyle='--', label=f"Température ambiante Ta={Ta}°C")

# Configuration des axes
plt.title("Loi de refroidissement de Newton : Champ de vecteurs et Solution", fontsize=14)
plt.xlabel("Temps t (min)")
plt.ylabel("Température T (°C)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.ylim(0, 100)
plt.show()
```

Considérons l’équation différentielle : $y'=x+y$ .

**a)** Montrer que $y(x)={{e}^{x}}-x-1$ est une solution particulière de l’E.D avec comme C.I. $y\left( 0 \right)=0$.

Remarque : 
  Cette E.D. peut-être résolue analytiquement avec la méthode de la variation des constantes.


**Solution**

a)  • $y'={{\left( {{e}^{x}}-x-1 \right)}^{'}}={{e}^{x}}-1$		
•  $x+y=x+\left( {{e}^{x}}-x-1 \right)={{e}^{x}}-1$
• $y(0)={{e}^{0}}-0-1=1-1=0$

**Remarque :   **

Résolution analytique de l’E.D. avec la méthode de la variation des constantes.

$y'=x+y\Leftrightarrow y'-y=x$     (la séparation des variables ne marche pas avec cette E.D.)

**Etape 01)**

Recherche de la solution générale pour l’équation sans second membre :$y'-y=0$

$y'-y=0\Leftrightarrow \frac{dy}{dx}-y=0\Leftrightarrow \frac{1}{y}\cdot dy=1\cdot dx\Leftrightarrow \int_{{}}^{{}}{\frac{1}{y}dy}=\int_{{}}^{{}}{1dx}\Leftrightarrow \ln \left| y \right|=x+CC\in \mathbb{R}$
 		
$\Leftrightarrow {{e}^{\ln \left| y \right|}}={{e}^{x+C}}\Leftrightarrow \left| y \right|={{e}^{x}}\cdot {{e}^{C}}\Leftrightarrow y=\underbrace{\pm {{e}^{C}}}_{\alpha }\cdot {{e}^{x}}\Leftrightarrow y=\alpha \cdot {{e}^{x}}\alpha \in \mathbb{R}$

**Etape 02)**

Recherche d’une solution générale pour l’équation avec second membre : $y'-y=x$     
On admet $\alpha =\alpha (x)$ et donc  est une fonction dépendante de x qui est telle que $y=\alpha (x)\cdot {{e}^{x}}$ soit solution de l’équation avec second membre.

$\left( \alpha \left( x \right)\cdot {{e}^{x}} \right)'-\left( \alpha \left( x \right)\cdot {{e}^{x}} \right)=x$ 				(substitution avec $y=\alpha \left( x \right)\cdot {{e}^{x}}$)

$\Leftrightarrow \alpha '\left( x \right)\cdot {{e}^{x}}+{\alpha \left( x \right)\cdot {{e}^{x}}}-{\alpha \left( x \right)\cdot {{e}^{x}}}=x$			(calcul différentiel) 


$\Leftrightarrow \alpha '\left( x \right)\cdot {{e}^{x}}=x$						(simplification) 


$\Leftrightarrow \alpha '\left( x \right)=x\cdot {{e}^{-x}}$						(simplification) 


$\Leftrightarrow \alpha \left( x \right)=\int_{{}}^{{}}{x}{{e}^{-x}}dx=-x{{e}^{-x}}-{{e}^{-x}}+CC\in \mathbb{R}$	(intégration par partie ) 


Finalement, la solution générale de l’équation avec second membre est :

$y=\alpha (x)\cdot {{e}^{x}}=\left( -x{{e}^{-x}}-{{e}^{-x}}+C \right)\cdot {{e}^{x}}=C\cdot {{e}^{x}}-x-1C\in \mathbb{R}$

 Avec comme condition initiale $y\left( 0 \right)=0\Leftrightarrow C\cdot {{e}^{0}}-0-1=0\Leftrightarrow C=1$

### Exercice 50

Loi de croissance logistique
Imaginons qu’une population donnée dans un milieu donné ne puisse pas dépasser une certaine quantité M. Les causes de cette limitation sont multiples ; prenons comme exemples le manque de logement dans une ville, le manque de nourriture sur une île, etc.

Pierre François Verlhust (1804 -1849 , mathématicien belge) exprima la vitesse d’accroissement d’une population à l’aide d’une équation différentielle du 1e ordre qui tienne compte du fait que  le milieu est limité et soit valable tout au long de la croissance (c’est-à-dire aussi bien au départ, 
quand rien n’arrête la reproduction que plus tard, quand le milieu est saturé) :

$\underbrace{\frac{dP}{dt}=k\cdot P}_{\begin{smallmatrix} 
 loidecroissance \\ 
 \exp onentielle 
\end{smallmatrix}}\cdot \underbrace{\frac{M-P}{M}}_{\begin{smallmatrix} 
 fonction \\ 
 retardatrice 
\end{smallmatrix}}$

où
- P est le nombre d’individus au temps t  d’où $P=P\left( t \right)$.
- k est le taux de croissance (constante).
- M est la capacité d’accueil du milieu (constante).

Avec cette équation différentielle nous arrivons à un bon compromis. 

En effet :

i) Lorsque la croissance débute, P est très petit en comparaisons de M : 

   $\frac{dP}{dt}=k\cdot P\cdot \frac{M-P}{M}\simeq k\cdot P\cdot \frac{M-0}{M}=k\cdot P$
   
  Ce qui est bien l’expression d’une croissance exponentielle : $\frac{dP}{dt}\simeq k\cdot P\Rightarrow P(t)\simeq {{P}_{0}}\cdot {{e}^{k\cdot t}}$
  
ii) Lorsque P est très proche de M donc en fin de croissance : $\frac{dP}{dt}=k\cdot P\cdot \frac{M-P}{M}\simeq k\cdot P\cdot \frac{M-M}{M}=0$

  Ce qui est bien l’expression d’une croissance nulle :  $\frac{dP}{dt}\simeq 0\Rightarrow P(t)\simeq Cte$ (droite horizontale)

**En conclusion**, la croissance d’une population se caractérise par une phase initiale de croissance exponentielle suivie d’une phase de ralentissement de la vitesse d’accroissement due aux facteurs limitants. 

**Schéma** comparant les lois de croissances exponentielles et logistiques :




**Enoncé**

**a)** Montrer que $P=\frac{M}{\left( \frac{M}{{{P}_{0}}}-1 \right)\cdot {{e}^{-k\cdot t}}+1}$ est une solution particulière de l’E.D avec comme C.I.  $P\left( 0 \right)={{P}_{0}}$ .

Remarque : Cette E.D. peut-être résolue analytiquement par séparation des variables.

**Solution**

  - $\frac{dP}{dt}={{\left( \frac{M}{\left( \frac{M}{{{P}_{0}}}-1 \right)\cdot {{e}^{-k\cdot t}}+1} \right)}^{'}}=M{{\left( {{\left( \left( \frac{M}{{{P}_{0}}}-1 \right)\cdot {{e}^{-k\cdot t}}+1 \right)}^{-1}} \right)}^{'}}$
  $=M\cdot \frac{-1}{{{\left( \left( \frac{M}{{{P}_{0}}}-1 \right)\cdot {{e}^{-k\cdot t}}+1 \right)}^{2}}}\cdot {{\left( \left( \frac{M}{{{P}_{0}}}-1 \right)\cdot {{e}^{-k\cdot t}}+1 \right)}^{'}}$ 
 $=M\cdot \frac{-1}{{{\left( \left( \frac{M}{{{P}_{0}}}-1 \right)\cdot {{e}^{-k\cdot t}}+1 \right)}^{2}}}\cdot \left( -k\left( \frac{M}{{{P}_{0}}}-1 \right)\cdot {{e}^{-k\cdot t}} \right)=\frac{kM\left( \frac{M}{{{P}_{0}}}-1 \right)\cdot {{e}^{-k\cdot t}}}{{{\left( \left( \frac{M}{{{P}_{0}}}-1 \right)\cdot {{e}^{-k\cdot t}}+1 \right)}^{2}}}$
 
  - $k\cdot P\cdot \frac{M-P}{M}=k\cdot \frac{M}{\left( \frac{M}{{{P}_{0}}}-1 \right)\cdot {{e}^{-k\cdot t}}+1}\cdot \frac{M-\left( \frac{M}{\left( \frac{M}{{{P}_{0}}}-1 \right)\cdot {{e}^{-k\cdot t}}+1} \right)}{M}$
  $=\frac{k}{\left( \frac{M}{{{P}_{0}}}-1 \right)\cdot {{e}^{-k\cdot t}}+1}\cdot \frac{M\left( \left( \frac{M}{{{P}_{0}}}-1 \right)\cdot {{e}^{-k\cdot t}}+1 \right)-M}{\left( \frac{M}{{{P}_{0}}}-1 \right)\cdot {{e}^{-k\cdot t}}+1}=\frac{kM\left( \frac{M}{{{P}_{0}}}-1 \right)\cdot {{e}^{-k\cdot t}}}{{{\left( \left( \frac{M}{{{P}_{0}}}-1 \right)\cdot {{e}^{-k\cdot t}}+1 \right)}^{2}}}$
                         
  - $P\left( 0 \right)=\frac{M}{\left( \frac{M}{{{P}_{0}}}-1 \right)\cdot {{e}^{-k\cdot 0}}+1}=\frac{M}{\left( \frac{M}{{{P}_{0}}}-1 \right)+1}=\frac{M}{\frac{M}{{{P}_{0}}}}={{P}_{0}}$

- Nom de la fonction : 	`RK4( f , x0 , y0 , b , n)`

```python
def f(x,y):
    return x+y

# Méthode Rk4 ( approximation yn )

def rk4(f,x0,y0,b,n):
    h=(b-x0)/n
    X, Y = [x0], [y0]
    for k in range(1,n+1):
        x1 = x0 + h
        k1 = h * f(x0,y0)
        k2 = h * f(x0+h/2, y0+k1/2)
        k3 = h * f(x0+h/2, y0+k2/2)
        k4 = h * f(x0+h, y0+k3)
        y1 = y0+(1/6)*(k1+2*k2+2*k3+k4)
        X = X + [x1]
        Y = Y + [y1]
        x0 = x1
        y0 = y1
    return X,Y
```

```python
from math import exp
import matplotlib.pyplot as plt

# Paramètres
x0,y0,xmax=0,0,3

# Solution exacte y(x)
def y(x):
    return exp(x)-x-1

def tab_val(y,x0,b,n):
    X, Y = [x0], [y(x0)]
    h=(b-x0)/n
    while x0 <= b:
        X = X + [x0]
        Y = Y + [y(x0)]
        x0 = x0 + h
    return X,Y

# Fonction f
def f(x,y):
    return x+y

# Méthode de résolutions
def euler(f,x0,y0,b,n):
    h=(b-x0)/n
    X, Y = [x0], [y0]
    for k in range(1,n+1):
        x1 = x0 + h
        y1 = y0 + h * f(x0,y0)
        X = X + [x1]
        Y = Y + [y1]
        x0 = x1
        y0 = y1
    return X,Y
def runge(f,x0,y0,b,n):
    h=(b-x0)/n
    X, Y = [x0], [y0]
    for k in range(1,n+1):
        x1=x0+h
        k1=h*f(x0,y0)
        k2=h*f(x0+h/2,y0+k1/2)
        y1=y0+k2
        X = X + [x1]
        Y = Y + [y1]
        x0 = x1
        y0 = y1
    return X, Y
def rk4(f,x0,y0,b,n):
    h=(b-x0)/n
    X, Y = [x0], [y0]
    for k in range(1,n+1):
        x1 = x0 + h
        k1 = h * f(x0,y0)
        k2 = h * f(x0+h/2, y0+k1/2)
        k3 = h * f(x0+h/2, y0+k2/2)
        k4 = h * f(x0+h, y0+k3)
        y1 = y0+(1/6)*(k1+2*k2+2*k3+k4)
        X = X + [x1]
        Y = Y + [y1]
        x0 = x1
        y0 = y1
    return X,Y

# calcul des solutions
X_euler, Y_euler = euler(f,x0,y0,xmax,2**2)
X_runge, Y_runge = runge(f,x0,y0,xmax,2**2)
X_rk4, Y_rk4 = rk4(f,x0,y0,xmax,2**2)
X_exact = X_euler
Y_exact = [y(x) for x in X_exact]

# affichage des solutions
plt.figure()
plt.grid(True)
plt.xlabel("x")
plt.ylabel("y")
plt.xlim(0, xmax)
p1,= plt.plot(X_exact, Y_exact, linestyle='-',color='black', marker='o',linewidth=4)
p2,= plt.plot(X_euler, X_euler, linestyle='--',color='red', marker='o',linewidth=2)
p3,= plt.plot(X_runge, Y_runge, linestyle='--',color='blue', marker='o',linewidth=2)
p4,= plt.plot(X_rk4, Y_rk4, linestyle='--',color='green', marker='o',linewidth=2)
plt.legend([p1,p2,p3, p4],['y(x) solution exacte','yn solution Euler (rouge)',
                       'yn solution Runge (bleu)', 'yn solution rk4 (vert)'])
plt.show()

```
